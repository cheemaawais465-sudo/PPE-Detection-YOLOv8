```python
from ultralytics import YOLO
import os
from collections import Counter

# Class names
class_names = {
    0: "Person",
    1: "Hardhat",
    2: "Safety Vest",
    3: "NO-Hardhat",
    4: "NO-Safety Vest"
}

label_folder = "/content/manual_99/test/labels"

print("Ground Truth Objects in 10 Test Images:\n")

for label_file in sorted(os.listdir(label_folder)):
    if not label_file.endswith(".txt"):
        continue

    path = os.path.join(label_folder, label_file)

    classes = []

    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 5:
                classes.append(int(parts[0]))

    counts = Counter(classes)

    readable = []
    for cls_id, count in counts.items():
        readable.append(f"{class_names.get(cls_id, str(cls_id))}: {count}")

    print(label_file.replace(".txt", ""))
    print("  " + ", ".join(readable))
```

```python
from collections import Counter
import os

all_splits = ["train", "valid", "test"]

error_scores = []

for split in all_splits:
    image_folder = f"/content/manual_99/{split}/images"
    label_folder = f"/content/manual_99/{split}/labels"

    for image_file in os.listdir(image_folder):
        if not image_file.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        image_path = os.path.join(image_folder, image_file)
        label_path = os.path.join(
            label_folder,
            os.path.splitext(image_file)[0] + ".txt"
        )

        # Ground truth
        gt = []
        with open(label_path, "r") as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) >= 5:
                    gt.append(int(parts[0]))

        # Prediction
        pred_result = model.predict(
            source=image_path,
            conf=0.25,
            verbose=False
        )[0]

        pred = pred_result.boxes.cls.cpu().numpy().astype(int).tolist()

        # Difference in number of detected objects
        difference = abs(len(gt) - len(pred))

        error_scores.append({
            "image": image_file,
            "split": split,
            "actual_objects": len(gt),
            "predicted_objects": len(pred),
            "difference": difference
        })

# Highest-error images first
error_scores = sorted(
    error_scores,
    key=lambda x: x["difference"],
    reverse=True
)

print("Top 15 potentially weak predictions:\n")

for i, item in enumerate(error_scores[:15], 1):
    print(
        f"{i}. {item['image']} | "
        f"Split: {item['split']} | "
        f"Actual: {item['actual_objects']} | "
        f"Predicted: {item['predicted_objects']} | "
        f"Difference: {item['difference']}"
    )
```

```python
from collections import Counter
import os
import glob

images_folder = "/content/manual_99/test/images"
labels_folder = "/content/manual_99/test/labels"

class_names = {
    0: "Person",
    1: "Hardhat",
    2: "Safety Vest",
    3: "NO-Hardhat",
    4: "NO-Safety Vest"
}

for img_path in sorted(glob.glob(images_folder + "/*")):
    base = os.path.splitext(os.path.basename(img_path))[0]
    label_path = os.path.join(labels_folder, base + ".txt")

    # Ground Truth
    gt = []
    if os.path.exists(label_path):
        with open(label_path, "r") as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    gt.append(int(parts[0]))

    # Prediction
    result = model.predict(img_path, conf=0.25, verbose=False)[0]
    pred = result.boxes.cls.cpu().numpy().astype(int).tolist()

    gt_count = Counter(gt)
    pred_count = Counter(pred)

    print("\n" + "="*70)
    print("IMAGE:", base)
    print("Ground Truth:", {class_names[k]: v for k, v in gt_count.items()})
    print("Prediction:  ", {class_names[k]: v for k, v in pred_count.items()})

    if len(pred) > len(gt):
        print("ERROR TYPE: Possible False Positives / Over-detection")
    elif len(pred) < len(gt):
        print("ERROR TYPE: Possible Missed Detection")
    elif set(pred) != set(gt):
        print("ERROR TYPE: Possible Wrong Class")
    else:
        print("STATUS: Counts look similar")
```
